# FinanceTracker: демонстрация CLI
Необязательно. Windows/Linux + Python/Jupyter + .NET 10 SDK. Расположите notebook в `examples` исходного проекта; выполните ячейки по порядку. Данные создаются во временной папке.

```python
from pathlib import Path
import subprocess, tempfile
root = Path.cwd().resolve()
if root.name == "examples": root = root.parent
project = root / "src/FinanceTracker.Cli/FinanceTracker.Cli.csproj"
assert project.exists(), "Откройте notebook из корня проекта или папки examples"
tmp = tempfile.TemporaryDirectory(prefix="finance-demo-")
data = str(Path(tmp.name) / "finance.json")
def cli(*args):
    result = subprocess.run(["dotnet", "run", "--project", str(project), "-c", "Release", "--", *args, "--data", data], capture_output=True, text=True, encoding="utf-8")
    print(result.stdout)
    if result.returncode: raise RuntimeError(result.stderr)
cli("demo", "--month", "2026-10")
```

```python
cli("list", "--month", "2026-10")
cli("summary", "--month", "2026-10")
```

```python
cli("add", "--type", "expense", "--amount", "350.50", "--category", "Продукты", "--date", "2026-10-12", "--note", "Обед")
cli("summary", "--month", "2026-10")
```

```python
cli("budget", "--month", "2026-10", "--amount", "20000")
cli("list", "--type", "expense", "--search", "обед")
cli("export", "--file", str(Path(tmp.name) / "report.csv"), "--month", "2026-10")
```
